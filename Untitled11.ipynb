```python
flores = {
    "tipo": ["rosa", "tulipan", "girasol", "amapola", "lirio"],
    "cantidad": [1, 17, 3, 8, 5],
    "precio": [4000, 80000, 5000, 15000, 23000]
}

carrito_flores = []
carrito_cantidades = []
carrito_subtotales = []

print("Bienvenido a mi floristeria\n")
print("Menú")

while True:
    print("\n¿Qué deseas hacer?")
    print("1. Consultar flores")
    print("2. Comprar flores")
    print("3. Pagar")
    print("4. Factura de compra")
    print("5. Salir")

    opc = int(input("Ingrese su opción: "))

    if opc == 1:
        print("\nEstas son nuestras flores disponibles:")
        print("Num | Flor | Cantidad | Precio")
        for i in range(len(flores["tipo"])):
            print(i + 1, "-", flores["tipo"][i],, flores["cantidad"][i], , flores["precio"][i])

    elif opc == 2:
        print("\n--- Comprar Flores ---")
        for i in range(len(flores["tipo"])):
            print(i + 1, ".", flores["tipo"][i], "(Stock:", flores["cantidad"][i], "- Precio: $", flores["precio"][i], ")")

        seleccion = int(input("Seleccione el número de la flor: ")) - 1

        if seleccion >= 0 and seleccion < len(flores["tipo"]):
            cant_deseada = int(input("¿Cuántas unidades desea? "))

            if cant_deseada <= flores["cantidad"][seleccion]:
                flores["cantidad"][seleccion] = flores["cantidad"][seleccion] - cant_deseada
                subtotal = cant_deseada * flores["precio"][seleccion]

                carrito_flores.append(flores["tipo"][seleccion])
                carrito_cantidades.append(cant_deseada)
                carrito_subtotales.append(subtotal)

                print("¡Agregado al carrito!")
            else:
                print("No hay suficiente stock disponible.")
        else:
            print("Selección no válida.")

    elif opc == 3:
        if len(carrito_flores) == 0:
            print("\nEl carrito está vacío.")
        else:
            total = sum(carrito_subtotales)
            print("\nEl total a pagar es: ",total)
            print("¡Gracias por su pago!")
            carrito_flores = []
            carrito_cantidades = []
            carrito_subtotales = []

    elif opc == 4:
        print("\n--- Factura de Compra ---")
        if len(carrito_flores) == 0:
            print("No tienes productos en el carrito.")
        else:
            for i in range(len(carrito_flores)):
                print(carrito_flores[i], "- Cantidad:", carrito_cantidades[i], "- Subtotal: $", carrito_subtotales[i])
            total = sum(carrito_subtotales)
            print("TOTAL:", total)

    elif opc == 5:
        print("Gracias por su visita. ¡Adiós!")
        break

    else:
        print("Opción no válida")
```

```python
flores = {
    "tipo": ["rosa", "tulipan", "girasol", "amapola", "lirio"],
    "cantidad": [1, 17, 3, 8, 5],
    "precio": [4000, 80000, 5000, 15000, 23000]
}

carrito_flores = []
carrito_cantidades = []
carrito_subtotales = []

print("Bienvenido a mi floristeria\n")
print("Menú")

while True:
    print("\n¿Qué deseas hacer?")
    print("1. Consultar flores")
    print("2. Comprar flores")
    print("3. Pagar")
    print("4. Factura de compra")
    print("5. Salir")

    opc = int(input("Ingrese su opción: "))

    if opc == 1:
        print("\nEstas son nuestras flores disponibles:")
        print("Num | Flor | Cantidad | Precio")
        for i in range(len(flores["tipo"])):
            print(i + 1, "-", flores["tipo"][i], "-", flores["cantidad"][i], "-", flores["precio"][i])

    elif opc == 2:
        print("\n--- Comprar Flores ---")
        for i in range(len(flores["tipo"])):
            print(i + 1, ".", flores["tipo"][i], "(Stock:", flores["cantidad"][i], "- Precio: $", flores["precio"][i], ")")

        seleccion = int(input("Seleccione el número de la flor: ")) - 1

        if seleccion >= 0 and seleccion < len(flores["tipo"]):
            cant_deseada = int(input("¿Cuántas unidades desea? "))

            if cant_deseada <= flores["cantidad"][seleccion]:
                flores["cantidad"][seleccion] = flores["cantidad"][seleccion] - cant_deseada
                subtotal = cant_deseada * flores["precio"][seleccion]

                carrito_flores.append(flores["tipo"][seleccion])
                carrito_cantidades.append(cant_deseada)
                carrito_subtotales.append(subtotal)

                print("¡Agregado al carrito!")
            else:
                print("No hay suficiente stock disponible.")
        else:
            print("Selección no válida.")

    elif opc == 3:
        if len(carrito_flores) == 0:
            print("\nEl carrito está vacío.")
        else:
            total = 0
            print("\n--- Procesando Pago ---")
            for flor, cant, sub in zip(carrito_flores, carrito_cantidades, carrito_subtotales):
                print("Pagando:", cant, "x", flor, "- Subtotal: $", sub)
                total = total + sub

            print("El total pagado es: $", total)
            print("¡Gracias por su compra!")
            carrito_flores = []
            carrito_cantidades = []
            carrito_subtotales = []

    elif opc == 4:
        print("\n--- Factura de Compra ---")
        if len(carrito_flores) == 0:
            print("No tienes productos en el carrito.")
        else:
            for i in range(len(carrito_flores)):
                print(carrito_flores[i], "- Cantidad:", carrito_cantidades[i], "- Subtotal: $", carrito_subtotales[i])
            total = sum(carrito_subtotales)
            print("TOTAL:", total)

    elif opc == 5:
        print("Gracias por su visita. ¡Adiós!")
        break

    else:
        print("Opción no válida")
```
