```python
#MASCOTA VIRTUAL
class mi_russa():
  def __init__(self, nombre, raza, energia):
    self.nombre = nombre
    self.raza = raza
    self.hambre = 50
    self.energia = energia

  def saludar(self):
    print(f"Здравствуйте, я- {self.nombre} ")

  def comer(self):
    if 0 <= self.hambre <= 25:
      print("🙂 im fineee xixixi")
    elif 25 < self.hambre <= 70:
      print("😄 Готовь кошелёк, мой прекрасный латиноамериканец.")
    elif 70 < self.hambre <= 90:
      print("😣 Я так голоден!")
    elif self.hambre > 90:
      print("👻Этот мудак не умел обо мне позаботиться.")

  def alimentar(self, c):
    comida = {
        "leche": 10,
        "polvo": 50
    }
    if c in comida:
      self.hambre = max(0, self.hambre - comida[c])
      self.energia += comida[c] / 2
      print(f"Le diste {c}. Hambre actual: {self.hambre}")
    else:
      print(f"No hay {c} disponible para alimentar.")

  def jugar(self):
    self.energia -= 5
    self.hambre = min(100, self.hambre + 15)
    print(f"{self.nombre} ha jugado.")
    print(f"Energia restante: {self.energia} | Hambre actual: {self.hambre}")
```

```python
russa1 = mi_russa("Валентина", "Rusa", 100)
```

```python
russa1.comer()
```
