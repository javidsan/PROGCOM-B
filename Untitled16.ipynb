```python
# Inventario de flores
flores = {
    "tipo": ["margarita", "azucena", "rosa"],
    "cant": [10, 20, 30],
    "precios": [1200, 3000, 1200]
}

# Carrito de compras del usuario
compra_user = {
    "tipo": [],
    "cantidad": [],
    "precio": []
}

chiste_evasores = "¡Ojo! No cobramos impuestos... ¡somos evasores de impuestos declarados! 🤫🤫"


# FUNCIÓN 1: Mostrar el inventario disponible
# Parámetro: flores (diccionario con el inventario)
# Qué hace: muestra las flores, las cantidades y sus precios.
def mostrar_inventario(flores):
    print("\n--- INVENTARIO DE FLORES EN STOCK ---")

    for i in range(len(flores["tipo"])):
        print(
            f"Flor: {flores['tipo'][i]} | "
            f"Stock: {flores['cant'][i]} | "
            f"Precio: ${flores['precios'][i]}"
        )


# FUNCIÓN 2: Comprar flores
# Parámetros: flores (inventario), compra_user (carrito).
# Qué hace: solicita una flor y una cantidad, verifica el stock,
# agrega la compra al carrito y descuenta las unidades disponibles.
def comprar_flores(flores, compra_user):
    print("\n--- COMPRAR FLORES ---")

    flor_opc = input("¿Qué flor quieres comprar?: ").lower().strip()

    if flor_opc in flores["tipo"]:
        idx = flores["tipo"].index(flor_opc)

        try:
            flor_cant = int(
                input(f"¿Cuántas unidades de {flor_opc} quieres?: ")
            )
        except ValueError:
            print("Debes ingresar una cantidad válida.")
            return

        if flor_cant <= 0:
            print("La cantidad debe ser mayor que cero.")

        elif flor_cant <= flores["cant"][idx]:
            compra_user["tipo"].append(flor_opc)
            compra_user["cantidad"].append(flor_cant)
            compra_user["precio"].append(flores["precios"][idx])

            flores["cant"][idx] -= flor_cant

            print(
                f"¡Listo! Has agregado {flor_cant} "
                f"{flor_opc}(s) a tu carrito."
            )
            print(f"Quedan {flores['cant'][idx]} {flor_opc}(s) en stock.")

        else:
            print("No tenemos suficiente cantidad de esa flor en stock.")

    else:
        print("Esa flor no está disponible en la tienda.")


# FUNCIÓN 3: Calcular el total de la compra
# Parámetro: compra_user (carrito con cantidades y precios).
# Qué hace: multiplica la cantidad por el precio de cada flor
# y devuelve el valor total que debe pagar el cliente.
def calcular_total(compra_user):
    total = 0

    for i in range(len(compra_user["tipo"])):
        total += (
            compra_user["cantidad"][i] *
            compra_user["precio"][i]
        )

    return total


# FUNCIÓN 4: Procesar el pago
# Parámetros: compra_user (carrito), chiste_evasores (mensaje final).
# Qué hace: calcula el total y permite pagar en efectivo o a cuotas.
def procesar_pago(compra_user, chiste_evasores):
    print("\n--- PROCESO DE PAGO ---")

    if len(compra_user["tipo"]) == 0:
        print("Tu carrito está vacío.")
        return

    total_pagar = calcular_total(compra_user)

    print(f"El total acumulado de tu compra es: ${total_pagar}")
    print("¿Cómo deseas realizar el pago?")
    print("1. Efectivo")
    print("2. Tarjeta (a cuotas)")

    pago_opc = input("Selecciona tu método (1 o 2): ")

    match pago_opc:
        case "1":
            try:
                plata_cliente = float(
                    input("¿Con cuánto dinero vas a pagar?: $")
                )
            except ValueError:
                print("Debes ingresar un valor válido.")
                return

            if plata_cliente >= total_pagar:
                cambio = plata_cliente - total_pagar
                print(f"¡Pago aceptado! Tu cambio es de: ${cambio:.2f}")
                print(chiste_evasores)
            else:
                falta = total_pagar - plata_cliente
                print(f"No te alcanza la plata. Te faltan: ${falta:.2f}")

        case "2":
            print("Opciones de cuotas disponibles: 1, 3, 6, 12 cuotas.")

            try:
                cuotas = int(
                    input("¿A cuántas cuotas
deseas
financiarlo?: ")
                )
            except ValueError:
                print("Debes ingresar un número de cuotas válido.")
                return

            if cuotas in [1, 3, 6, 12]:
                valor_cuota = total_pagar / cuotas
                print(
                    f"¡Pago aprobado! Pagarás {cuotas} cuota(s) "
                    f"de: ${valor_cuota:.2f} cada una."
                )
                print(chiste_evasores)
            else:
                print("La cantidad de cuotas seleccionada no es válida.")

        case _:
            print("Método de pago no válido.")


# FUNCIÓN 5: Generar la factura
# Parámetro: compra_user (carrito de compras).
# Qué hace: muestra cada producto comprado, su cantidad,
# el precio unitario y el total de la factura.
def generar_factura(compra_user):
    print("\n--- FACTURA DE COMPRA ---")

    if len(compra_user["tipo"]) == 0:
        print("No hay productos para generar la factura.")
        return

    for i in range(len(compra_user["tipo"])):
        flor = compra_user["tipo"][i]
        cantidad = compra_user["cantidad"][i]
        precio = compra_user["precio"][i]
        subtotal = cantidad * precio

        print(
            f"Flor: {flor} | Cantidad: {cantidad} | "
            f"Precio unitario: ${precio} | Subtotal: ${subtotal}"
        )

    total = calcular_total(compra_user)
    print(f"TOTAL A PAGAR: ${total}")


# FUNCIÓN 6: Ejecutar el programa principal
# Parámetros: ninguno.
# Qué hace: muestra el menú y llama a las funciones
# correspondientes según la opción elegida.
def principal():
    print("Bienvenido a mi floristería")

    while True:
        mostrar_inventario(flores)

        print("\n¿Qué desea hacer?")
        print("1. Comprar flores")
        print("2. Pagar")
        print("3. Generar factura")
        print("4. Salir")

        opc = input("Seleccione su opción (1, 2, 3 o 4): ")

        match opc:
            case "1":
                comprar_flores(flores, compra_user)

            case "2":
                procesar_pago(compra_user, chiste_evasores)

            case "3":
                generar_factura(compra_user)

            case "4":
                print("¡Gracias por visitar nuestra floristería!")
                break

            case _:
                print("Opción no válida. Intenta nuevamente.")
```
