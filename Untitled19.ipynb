```python
#MASCOTA VIRTUAL
class mi_russa():
  def __init__(self, nombre, raza, energia):
    self.nombre = nombre
    self.raza = raza
    self.hambre = 50
    self.energia = energia

  def saludar(self):
    print(f"Здравствуйте, я- {self.nombre} ")

  def comer(self):
    comida = {
        "leche": 10,
        "polvo": 50
    }
    if self.hambre > 0 or self.hambre <= 25:
      print("🙂")
    elif self.hambre > 25 and self.hambre <= 70:
      print("😄")
    elif self.hambre > 70 and self.hambre <= 90:
      print("😣")
    elif self.hambre > 90 and self.hambre <= 100:
      print("👻")

  def alimentar(self, c):
    comida = {
        "leche": 10,
        "polvo": 50
    }
    if c in comida:
      self.hambre -= comida[c]
      self.energia += comida[c] / 2
    else:
      print(f"No hay {c} disponible para alimentar.")

  def jugar(self):
    self.energia -= 5
    print(f"{self.nombre} ha jugado.\nEnergia restante: {self.energia}")
```

```python
russa1 = mi_russa("Валентина", "Rusa", 100)
```
