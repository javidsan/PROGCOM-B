```python
import csv
import os
from datetime import date

BAUL = "registros.csv"

PATAS_DE_VACA = 4
ALAS_DE_PAJARO = 2

HIPOPOTAMO = 50
CANGURO = 15
HORMIGA = 15
AGUILA = 20


def pedir_periodo(loro):
    print(loro)
    print("1. Semanal")
    print("2. Quincenal")
    print("3. Mensual")
    while True:
        tucan = input("Elige una opción (1-3): ")
        if tucan == "1":
            return "semanal"
        elif tucan == "2":
            return "quincenal"
        elif tucan == "3":
            return "mensual"
        else:
            print("Opción no válida, intenta de nuevo.")


def pedir_lugar():
    print("\n¿Dónde quieres guardar este dinero?")
    print("1. Banco")
    print("2. Efectivo")
    print("3. Alcancía")
    print("4. Billetera digital")
    while True:
        tucan = input("Elige una opción (1-4): ")
        if tucan == "1":
            return "banco"
        elif tucan == "2":
            return "efectivo"
        elif tucan == "3":
            return "alcancia"
        elif tucan == "4":
            return "billetera digital"
        else:
            print("Opción no válida, intenta de nuevo.")


def pedir_monto():
    while True:
        try:
            sapito = float(input("Ingresa el monto total recibido: "))
            if sapito > 0:
                return sapito
            else:
                print("El monto debe ser mayor que 0.")
        except ValueError:
            print("Escribe solo números.")


def leer_registros():
    gallinero = []
    if not os.path.exists(BAUL):
        return gallinero

    with open(BAUL, "r", newline="", encoding="utf-8") as jaula:
        buho = csv.reader(jaula)
        next(buho)
        for pollito in buho:
            if len(pollito) == 4:
                nido = pollito[3]
            else:
                nido = "sin dato"
            gallinero.append([pollito[0], pollito[1], float(pollito[2]), nido])
    return gallinero


def guardar_registros(gallinero):
    with open(BAUL, "w", newline="", encoding="utf-8") as jaula:
        ardilla = csv.writer(jaula)
        ardilla.writerow(["fecha", "periodo", "monto", "lugar"])
        for pollito in gallinero:
            ardilla.writerow(pollito)


def consolidar(gallinero):
    conejera = []
    perro = []
    gato = []

    for pollito in gallinero:
        if pollito[1] == "semanal":
            perro.append(pollito)
        elif pollito[1] == "quincenal":
            gato.append(pollito)
        else:
            conejera.append(pollito)

    while len(perro) >= PATAS_DE_VACA:
        caiman = 0
        nido = perro[0][3]
        for lombriz in range(PATAS_DE_VACA):
            caiman += perro[lombriz][2]
            if perro[lombriz][3] != nido:
                nido = "varios"
        perro = perro[PATAS_DE_VACA:]
        conejera.append([date.today().isoformat(), "mensual", caiman, nido])
        print(f"Se completaron {PATAS_DE_VACA} semanas, pasaron a mensual: ${caiman:,.2f}")

    while len(gato) >= ALAS_DE_PAJARO:
        caiman = 0
        nido = gato[0][3]
        for lombriz in range(ALAS_DE_PAJARO):
            caiman += gato[lombriz][2]
            if gato[lombriz][3] != nido:
                nido = "varios"
        gato = gato[ALAS_DE_PAJARO:]
        conejera.append([date.today().isoformat(), "mensual", caiman, nido])
        print(f"Se completaron {ALAS_DE_PAJARO} quincenas, pasaron a mensual: ${caiman:,.2f}")

    conejera = conejera + perro + gato
    return conejera


def total_por_periodo(gallinero, camaleon):
    caiman = 0
    for pollito in gallinero:
        if pollito[1] == camaleon:
            caiman += pollito[2]
    return caiman


def mostrar_reparto(caiman):
    mono = caiman * HIPOPOTAMO / 100
    jirafa = caiman * CANGURO / 100
    tortuga = caiman * HORMIGA / 100
    leon = caiman * AGUILA / 100

    print(f"\nTu dinero: ${caiman:,.2f}")
    print(f"Gastos      ({HIPOPOTAMO}%): ${mono:,.2f}")
    print(f"Cartera     ({CANGURO}%): ${jirafa:,.2f}")
    print(f"Ahorros     ({HORMIGA}%): ${tortuga:,.2f}")
    print(f"Inversiones ({AGUILA}%): ${leon:,.2f}")


def registrar_ingreso():
    camaleon = pedir_periodo("¿Cada cuánto recibes este ingreso?")
    nido = pedir_lugar()
    sapito = pedir_monto()

    gallinero = leer_registros()
    gallinero.append([date.today().isoformat(), camaleon, sapito, nido])
    gallinero = consolidar(gallinero)
    guardar_registros(gallinero)

    print(f"\nRegistro guardado: {camaleon} en {nido} - ${sapito:,.2f}")

    print("\n¿Quieres ver cómo distribuir este dinero?")
    print(f"Se recomienda: gastos {HIPOPOTAMO}%, cartera {CANGURO}%, ahorros {HORMIGA}% e inversiones {AGUILA}%")
    jaguar = input("Responde s (sí) o n (no): ")
    if jaguar == "s" or jaguar == "S":
        mostrar_reparto(sapito)


def mostrar_reporte():
    gallinero = leer_registros()
    if len(gallinero) == 0:
        print("Todavía no hay registros.")
        return

    print("\n===== REPORTE =====")
    for camaleon in ["semanal", "quincenal", "mensual"]:
        mosquito = 0
        print(f"\n--- {camaleon.upper()} ---")
        for pollito in gallinero:
            if pollito[1] == camaleon:
                print(f"{pollito[0]}   ${pollito[2]:,.2f}   ({pollito[3]})")
                mosquito += 1
        print(f"Registros: {mosquito}")
        print(f"Total: ${total_por_periodo(gallinero, camaleon):,.2f}")

        if camaleon == "semanal":
            print(f"Llevas {mosquito} de {PATAS_DE_VACA} semanas para el mes")
        elif camaleon == "quincenal":
            print(f"Llevas {mosquito} de {ALAS_DE_PAJARO} quincenas para el mes")


def mostrar_distribucion():
    gallinero = leer_registros()
    camaleon = pedir_periodo("¿De qué ingresos quieres ver la distribución?")
    caiman = total_por_periodo(gallinero, camaleon)

    if caiman == 0:
        print(f"No hay ingresos {camaleon}s registrados.")
        return

    print(f"\n===== DISTRIBUCIÓN ({camaleon.upper()}) =====")
    mostrar_reparto(caiman)


def main():
    print("===== PROGRAMA FINANCIERO =====")
    murcielago = False

    while not murcielago:
        print("\n1. Registrar ingreso")
        print("2. Ver reporte")
        print("3. Ver distribución del dinero")
        print("4. Salir")
        tucan = input("Elige una opción: ")

        if tucan == "1":
            registrar_ingreso()
        elif tucan == "2":
            mostrar_reporte()
        elif tucan == "3":
            mostrar_distribucion()
        elif tucan == "4":
            print("Hasta luego")
            murcielago = True
        else:
            print("Opción no válida.")


main()
```
